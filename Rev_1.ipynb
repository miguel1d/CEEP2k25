# Revisão de Variáveis e Tipos de Dados

## Tipos Básicos

```python
# int: números inteiros (sem casas decimais)
idade = 17
print(f"Idade: {idade}, Tipo: {type(idade)}")
```

```python
# float: ponto flutuante (com casas decimais)
altura = 1.80
print(f"Altura: {altura}, Tipo: {type(altura)}")
```

```python
# string: (ou str) texto puro
nome = "Abner Sales Pitondo Redondo"
print(f"Nome: {nome}, Tipo: {type(nome)}")
```

```python
# bool: booleano (sim, nao, verdade, mentira, true, false)
estudante = True
print(f"Estudante: {estudante}, Tipo: {type(estudante)}")
```

## Tipos de Coleção

```python
# lista: coleção ordenada e mutável
frutas = ["maça", "banana", "uva", "pera"]
print(f"Frutas: {frutas}, Tipo: {type(frutas)}")
```

```python
# tupla: coleção ordenada e imutável, geralmente usada para guardar coordenadas
cianorte = (-23.6631, -52.6054)
print(f"Coordenadas de Cianorte: {cianorte}, Tipo: {type(cianorte)}")
```

```python
# set: coleção não ordenada de itens únicos
numeros_unicos = {1, 2, 3, 2, 1}
print(f"Números únicos (set): {numeros_unicos}, Tipo: {type(numeros_unicos)}")
```
